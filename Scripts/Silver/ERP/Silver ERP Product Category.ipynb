# Silver Layer — ERP: Product Category (px_cat_g1v2)

Transforms the bronze `workspace.bronze.px_cat_g1v2` table into `workspace.silver.product_info2` with meaningful, readable column names and trimmed data.

---

## Table of Contents — Transformations Applied

| # | Transformation | Description |
|---|---|---|
| 1 | **Column Renaming** | Renamed abbreviated columns to meaningful names (`ID` -> `category_id`, `CAT` -> `category`, `SUBCAT` -> `subcategory`, `MAINTENANCE` -> `maintenance_required`) |
| 2 | **Trim Whitespace** | Trimmed leading/trailing whitespace from all string columns as a precautionary cleanup |

---

## Column Mapping (Bronze to Silver)

| Bronze Column | Silver Column | Transformation Applied |
|---|---|---|
| `ID` | `category_id` | Renamed + Trimmed |
| `CAT` | `category` | Renamed + Trimmed |
| `SUBCAT` | `subcategory` | Renamed + Trimmed |
| `MAINTENANCE` | `maintenance_required` | Renamed + Trimmed |

---

## Data Overview

| Property | Value |
|---|---|
| Total Rows | 37 |
| Distinct Categories | 4 (Accessories, Bikes, Clothing, Components) |
| Null Counts | 0 across all columns |
| Maintenance Values | `Yes`, `No` |

---

## Implementation

All transformations are implemented in **PySpark** (`pyspark.sql.functions`):
- `col()` — select and alias columns
- `trim()` — remove leading/trailing whitespace
- `df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable()` — write to silver layer

```python
from pyspark.sql.functions import col, trim

df = (spark.read.table("workspace.bronze.px_cat_g1v2")
    .select(
        trim(col("ID")).alias("category_id"),
        trim(col("CAT")).alias("category"),
        trim(col("SUBCAT")).alias("subcategory"),
        trim(col("MAINTENANCE")).alias("maintenance_required"),
    )
)

df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.silver.product_info2")
```
